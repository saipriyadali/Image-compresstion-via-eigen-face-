```python

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_olivetti_faces
from sklearn.decomposition import PCA

print("Libraries imported successfully!")
```

```python

# Load Olivetti Faces dataset
faces = fetch_olivetti_faces(shuffle=True, random_state=42)

X = faces.data
images = faces.images

print("Dataset loaded successfully!")
print("Total images:", len(images))
print("Image size:", images.shape[1], "x", images.shape[2])
```

```python

# Select first 5 face images
n_faces = 5

selected_images = images[:n_faces]
selected_vectors = X[:n_faces]

print("Number of selected faces:", len(selected_images))
print("Each image size:", selected_images[0].shape)
print("Flattened vector size:", selected_vectors[0].shape)
```

```python

plt.figure(figsize=(12, 3))

for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(selected_images[i], cmap='gray')
    plt.title(f"Face {i+1}")
    plt.axis("off")

plt.suptitle("Original 5 Face Images")
plt.show()
```

```python

# Flatten each 64 x 64 image into a 4096-dimensional vector
face_vectors = selected_images.reshape(5, -1)

print("Original image shape:", selected_images.shape)
print("Flattened data shape:", face_vectors.shape)
```

```python

# Calculate mean face
mean_face = np.mean(face_vectors, axis=0)

print("Mean face vector shape:", mean_face.shape)
```

```python

plt.figure(figsize=(4, 4))

plt.imshow(mean_face.reshape(64, 64), cmap='gray')
plt.title("Mean Face")
plt.axis("off")

plt.show()
```

```python

# Center the data
centered_faces = face_vectors - mean_face

print("Centered data shape:", centered_faces.shape)
```

```python

# Calculate covariance matrix
covariance_matrix = np.cov(centered_faces, rowvar=False)

print("Covariance matrix shape:", covariance_matrix.shape)
```

```python

# Singular Value Decomposition
U, S, Vt = np.linalg.svd(centered_faces, full_matrices=False)

# Top 3 eigenfaces
top_3_eigenfaces = Vt[:3]

print("Top 3 eigenfaces shape:", top_3_eigenfaces.shape)
```

```python

# Calculate eigenvalues from singular values
eigenvalues = (S ** 2) / (len(face_vectors) - 1)

print("Top 3 Eigenvalues:")
for i in range(3):
    print(f"Eigenvalue {i+1}: {eigenvalues[i]}")
```

```python

plt.figure(figsize=(12, 4))

for i in range(3):
    plt.subplot(1, 3, i + 1)

    eigenface = top_3_eigenfaces[i].reshape(64, 64)

    plt.imshow(eigenface, cmap='gray')
    plt.title(f"Eigenface {i+1}")
    plt.axis("off")

plt.suptitle("Top 3 Eigenfaces")
plt.show()
```

```python

# Project centered faces onto the top 3 eigenfaces
compressed_data = centered_faces @ top_3_eigenfaces.T

print("Original data shape:", face_vectors.shape)
print("Compressed data shape:", compressed_data.shape)
```

```python

# Reconstruct images from only 3 eigenfaces
reconstructed_faces = (
    compressed_data @ top_3_eigenfaces
) + mean_face

# Keep pixel values between 0 and 1
reconstructed_faces = np.clip(reconstructed_faces, 0, 1)

print("Reconstructed data shape:", reconstructed_faces.shape)
```

```python

plt.figure(figsize=(12, 6))

for i in range(5):

    # Original image
    plt.subplot(2, 5, i + 1)
    plt.imshow(selected_images[i], cmap='gray')
    plt.title(f"Original {i+1}")
    plt.axis("off")

    # Reconstructed image
    plt.subplot(2, 5, i + 6)
    plt.imshow(reconstructed_faces[i].reshape(64, 64), cmap='gray')
    plt.title(f"Compressed {i+1}")
    plt.axis("off")

plt.suptitle("Eigenface Image Compression: Original vs Reconstructed")
plt.tight_layout()
plt.show()
```

```python

# Original representation
original_values = 4096

# Compressed representation
compressed_values = 3

compression_ratio = original_values / compressed_values

print("Original values per image:", original_values)
print("Compressed values per image:", compressed_values)
print("Compression ratio:", round(compression_ratio, 2), ": 1")
```
